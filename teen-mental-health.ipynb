# Analysis of Mental Health in Teenagers

 The goal is to explore teenagers' daily habits and their relationship to indicators such as stress, anxiety, and depression.

## 1. Importing Libraries and Data

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

# Loading the dataset
data = pd.read_csv('Teen_Mental_Health_Dataset.csv')
```

## 2. Exploración Inicial (Estructura del Dataset)

```python
print("First entries:")
display(data.head())

print("\nLast entries:")
display(data.tail())
```

```python
print("General information:")
data.info()

print("\nDescriptive statistics:")
display(data.describe())
```

## 3. Visualización de Datos

```python
# Histograms of Numeric Variables
data.hist(figsize=(15, 10), bins=20, color='skyblue', edgecolor='black')
plt.suptitle('Variable Distribution', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
# Relationship Between Hours of Sleep and Stress Levels
plt.figure(figsize=(10, 6))
sns.scatterplot(x='sleep_hours', y='stress_level', data=data, alpha=0.2, color='coral')
plt.title('Relationship: Hours of Sleep vs Stress Level')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

No visible correlation is observed between the two variables. This indicates that sleep hours alone are not sufficient to predict an individual's stress level, and that additional variables are needed to predict the stress level with a certain degree of accuracy—perhaps even variables that are not present in the dataset.

## 4. Data Preprocessing

We convert categorical variables to numerical values so they can be included in the correlation analysis and the model.

```python
# Handling Null Values
null_data = data.dropna()

# Encoding Text Variables and Mapping Interaction Levels
mapping_social = {'low': 0, 'medium': 1, 'high': 2}
null_data['social_interaction_level'] = null_data['social_interaction_level'].map(mapping_social)

dummies_platform = pd.get_dummies(null_data['platform_usage'], drop_first=True, dtype=int)
null_data = pd.concat([null_data.drop('platform_usage', axis=1), dummies_platform], axis=1)

dummies_gender = pd.get_dummies(null_data['gender'], drop_first=True, dtype=int)
processed_data = pd.concat([null_data.drop('gender', axis=1), dummies_gender], axis=1)

processed_data.head()
```

## 5. Correlation Analysis

We generate a heatmap to identify which variables influence each other the most.

```python
plt.figure(figsize=(12, 10))
sns.heatmap(processed_data.corr(), annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation heatmap')
plt.show()
```

There does not appear to be a strong correlation between most of the variables.

## 6. Prediction Model (Linear Regression)

We train a model to predict stress levels based on daily habits.

```python
# Input variables (X) and target variable (Y)
X = processed_data[['sleep_hours', 'daily_social_media_hours', 'physical_activity']]
y = processed_data['stress_level']

stress_model = LinearRegression()
stress_model.fit(X, y)

print("Stress Model Successfully Trained.")

# Prediction example: 7 hours of sleep, 4 hours on social media, 1 hour of exercise
example = [[7.0, 4.0, 1.0]]
prediction = stress_model.predict(example)[0]
print(f"Prediction example: A user with these habits would have a stress level of: {prediction:.2f}/10")
```

# 7. Evaluate the effectiveness of the prediction

The data is split so that one part is used for training and the other for testing the prediction

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20)
```

Now we need to train the model using only the training data

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

Next, we make predictions using the test data (the 20%)

```python
prediction = model.predict(X_test)
```

We compare the predictions with the actual data. R² (r squared) is a number between 0 and 1 that indicates how well the predictions align with the actual data. A number close to zero indicates that the variables used in the model are not sufficient to predict another variable (in this case, the stress level).

```python
comparison = r2_score(y_test, prediction)
```

The corr() function could also be used, but I found it simpler to use the r2 function for a direct comparison.

Let's look at the comparison.

```python
print(f"Punctuation R²: {comparison:.2f} ({(comparison*100):.1f}%)")
```

In this case, the result is a negative R², indicating that there is no linear relationship between the input variables and the stress level, or that the model does not fit the data well.
Therefore, the stress level cannot be explained solely by hours of sleep, social media usage, and physical activity, suggesting that other factors are at play.
